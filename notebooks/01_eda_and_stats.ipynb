## Setup

Run `pip install -r requirements-training.txt` from the repository root, then select that Python environment as this notebook's kernel. The loader finds `data/raw` whether Jupyter was started from the repository root or the `notebooks` directory.


```python
from pathlib import Path
import pandas as pd

# Resolve the data directory from either common Jupyter working directory.
_candidates = [Path.cwd() / 'data' / 'raw', Path.cwd().parent / 'data' / 'raw']
DATA_DIR = next((p.resolve() for p in _candidates if p.is_dir()), None)
if DATA_DIR is None:
    raise FileNotFoundError('Could not find data/raw. Start Jupyter from the repository root or notebooks folder.')

required_files = {
    'loan_approval': 'loan_approval_dataset.csv',
    'loan_default': 'Loan_default.csv',
    'lending': 'loans_2007.csv',
}
missing = [filename for filename in required_files.values() if not (DATA_DIR / filename).is_file()]
if missing:
    raise FileNotFoundError('Missing dataset(s) in {}: {}'.format(DATA_DIR, ', '.join(missing)))

df_approval = pd.read_csv(DATA_DIR / required_files['loan_approval'])
df_loandefault = pd.read_csv(DATA_DIR / required_files['loan_default'])
df_lending = pd.read_csv(DATA_DIR / required_files['lending'])
print(f'Loaded datasets from {DATA_DIR}')
print(f'Approval: {df_approval.shape}; default: {df_loandefault.shape}; lending: {df_lending.shape}')
```

```python
# Basic inspection for the loan approval dataset
print('Rows and columns:', df_approval.shape)
display(df_approval.head())
display(df_approval.isna().sum().sort_values(ascending=False).head(10).to_frame('missing_values'))

# The source CSV has whitespace in several column names; normalize them once for analysis.
df_approval.columns = df_approval.columns.str.strip()
if 'loan_status' in df_approval.columns:
    display(df_approval['loan_status'].value_counts(dropna=False).to_frame('count'))
display(df_approval.describe(include='all').T.head(20))
```
